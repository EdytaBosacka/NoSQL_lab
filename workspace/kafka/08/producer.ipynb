```python
from kafka import KafkaProducer

producer = KafkaProducer(bootstrap_servers='kafka_streaming_lab:9092')

# wyślij do konkretnych partycji
producer.send('lab8-topic', value=b'msg-0', partition=0)
producer.send('lab8-topic', value=b'msg-1', partition=1)
producer.send('lab8-topic', value=b'msg-2', partition=2)

producer.flush()
producer.close()
```

```python
from kafka import KafkaConsumer, TopicPartition

consumer = KafkaConsumer(bootstrap_servers='kafka_streaming_lab:9092')

tps = [
    TopicPartition('lab8-topic', 0),
    TopicPartition('lab8-topic', 2)
]

consumer.assign(tps)

consumer.seek_to_beginning(*tps)

for msg in consumer:
    print(msg.partition, msg.offset, msg.value)
```

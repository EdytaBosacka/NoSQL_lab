```python
from kafka import KafkaConsumer, TopicPartition

consumer = KafkaConsumer(bootstrap_servers='kafka_streaming_lab:9092')

tp = TopicPartition('lab8-topic', 1)  # wybierasz partycję 1
consumer.assign([tp])

# opcjonalnie: od początku partycji
consumer.seek_to_beginning(tp)

for msg in consumer:
    print(
        f"partition={msg.partition}, offset={msg.offset}, value={msg.value}"
    )
```

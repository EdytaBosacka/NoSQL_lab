```python
from kafka import KafkaProducer
import json
import time

producer = KafkaProducer(
    bootstrap_servers='9092:9092',
    value_serializer=lambda v: json.dumps(v).encode('utf-8')
)

topic = 'lab2-topic'

for i in range(50):
    data = {
        "id": i,
        "temperature": round(20 + 5 * i, 2),
        "timestamp": time.time()
    }
    producer.send(topic, value=data)
    print(f"Sent: {data}")
    time.sleep(1)

producer.flush()
producer.close()
```

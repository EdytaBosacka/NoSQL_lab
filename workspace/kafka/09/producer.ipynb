```python
from kafka import KafkaProducer

producer = KafkaProducer(bootstrap_servers='kafka_streaming_lab:9092')

# wszystkie wiadomości z tym samym key trafią do tej samej partycji
producer.send('lab9-topic', key=b'user1', value=b'login')
producer.send('lab9-topic', key=b'user1', value=b'click')
producer.send('lab9-topic', key=b'user2', value=b'login')

producer.flush()
producer.close()
```

```python
producer = KafkaProducer(bootstrap_servers='kafka_streaming_lab:9092')

for i in range(25):
    producer.send('lab9-topic', key=b'user1', value=f'event-{i}'.encode())

producer.flush()
producer.close()
```
